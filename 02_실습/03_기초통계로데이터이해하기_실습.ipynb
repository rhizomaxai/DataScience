## B1. 중심: 평균·중앙값·최빈값

▶️ 키의 평균(mean)과 중앙값(median)을 구한다.

## B2. 퍼짐: 범위·분산·표준편차·사분위수

▶️ 키의 최솟값·최댓값·범위(최댓값 − 최솟값)를 구한다.

```python
print('최소:', df['키'].min(), '최대:', df['키'].max())
print('범위:', df['키'].max() - df['키'].min())
```

▶️ 분산(var)과 표준편차(std): 값들이 평균에서 평균적으로 얼마나 떨어져 있는지

```python
print('분산:', round(df['키'].var(), 1))
print('표준편차:', round(df['키'].std(), 1))
```

✏️ 빈칸 채우기: 키의 사분위수(25%, 50%, 75%)를 `quantile`로 구한다.

```python
df['키'].quantile(___)
```

### 오늘의 정리
- **구조 확인**: `shape` · `info()` · `nunique()` · `unique()` — 수치형과 범주형 구분
- **중심**: `mean()` · `median()` · `mode()` — 이상치가 있으면 중앙값
- **퍼짐**: `max()-min()` · `var()` · `std()` · `quantile()` · `describe()`
- **범주형 분포**: `value_counts()` · `value_counts(normalize=True)`
- **정렬**: `sort_values(by=..., ascending=...)`
- **그룹 비교**: `groupby(기준)[열].mean()` · `.agg([...])` — 인원수도 함께 보기
