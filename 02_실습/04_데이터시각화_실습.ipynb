## B1. 선 그래프: 시간에 따른 변화

▶️ 월별 평균 대여건수를 구한 뒤 선 그래프로 그린다. `plt.plot(x, y)` → `plt.show()`

```python
monthly = bike.groupby('월')['대여건수'].mean()
plt.plot(monthly.index, monthly.values)
plt.show()
```

✏️ 빈칸 채우기: 제목, x축·y축 이름을 붙인다. (좋은 그래프의 기본!)

```python
plt.plot(monthly.index, monthly.values)
plt.___('월별 하루 평균 대여건수')
plt.___('월')
plt.___('대여건수(건)')
plt.show()
```

🔧 수정: 마커·색·선 모양·격자를 바꿔 본다. `marker`('o','s','^'), `color`('tomato','teal'), `linestyle`('-','--',':')

```python
plt.plot(monthly.index, monthly.values, marker='o', color='teal', linestyle='--')
plt.title('월별 하루 평균 대여건수')
plt.xlabel('월')
plt.ylabel('대여건수(건)')
plt.xticks(range(1, 13))   # x축 눈금을 1~12월로
plt.grid(True)
plt.show()
```

✏️ 빈칸 채우기: 평일과 주말을 두 개의 선으로 그리고 범례(`label`, `legend`)를 단다.

```python
weekday = bike[bike['주말여부'] == '평일'].groupby('월')['대여건수'].mean()
weekend = bike[bike['주말여부'] == '주말'].groupby('월')['대여건수'].mean()
plt.plot(weekday.index, weekday.values, marker='o', label=___)
plt.plot(weekend.index, weekend.values, marker='s', label=___)
plt.___
plt.title('평일·주말 월별 평균 대여건수')
plt.xlabel('월')
plt.ylabel('대여건수(건)')
plt.show()
```

## B2. 막대 그래프: 항목끼리 비교

# PART C. seaborn — 표(DataFrame)를 주면 알아서 그린다
공통 문법: `sns.그래프종류(data=표, x='열', y='열', hue='색으로 나눌 열')`

▶️ countplot: `value_counts` 없이 바로 센다. `hue`로 성별을 색으로 나눈다.

```python
sns.countplot(data=df, x='혈액형', hue='성별')
plt.title('혈액형별 인원 (성별)')
plt.show()
```

✏️ 빈칸 채우기: barplot은 그룹별 **평균**을 막대로 그린다. (3차시 groupby와 같은 일) 성별 평균 운동시간을 그린다.

```python
sns.barplot(data=df, x=___, y=___, errorbar=None)
plt.title('성별 평균 주당 운동시간')
plt.ylabel('시간')
plt.show()
```

👀 관찰: `errorbar=None`을 지우고 실행해 보자. 막대 위의 검은 선(오차 막대)은 '평균이 이 정도 범위에서 흔들릴 수 있다'는 표시다.

```python
sns.barplot(data=df, x='성별', y='운동시간')
plt.title('성별 평균 주당 운동시간 (오차 막대 포함)')
plt.show()
```

### 오늘의 정리
| 목적 | 차트 | matplotlib | seaborn | 판다스 |
|---|---|---|---|---|
| 시간에 따른 변화 | 선 | `plt.plot` | `sns.lineplot` | `.plot()` |
| 항목 비교 | 막대 | `plt.bar` / `barh` | `sns.barplot` · `sns.countplot` | `.plot.bar()` |
| 분포 | 히스토그램 · 상자 | `plt.hist` · `plt.boxplot` | `sns.histplot` · `sns.boxplot` | `.plot.hist()` |
| 관계 | 산점도 · 히트맵 | `plt.scatter` | `sns.scatterplot` · `sns.heatmap` | `.plot.scatter()` |
| 구성비 | 원 (항목 적을 때만) | `plt.pie` | — | `.plot.pie()` |

좋은 그래프의 기본: **제목 · 축 이름과 단위 · 세로축 0부터(막대) · 출처**
